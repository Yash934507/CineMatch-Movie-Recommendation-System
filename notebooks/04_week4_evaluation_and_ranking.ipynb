# CineMatch - Week 4: Final Evaluation and Ranking
This notebook compares rating accuracy and ranked recommendation quality, selects the final model from measured results, and documents the cold-start strategy.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
summary = json.loads((ROOT / 'data' / 'processed' / 'week4_summary.json').read_text())
```

## 1. Protocol
Rating error uses the same deterministic 2,000 held-out ratings for all models. Ranking uses 100 seeded users who have at least one held-out rating >= 4.0. Each model scores the same held-out movies plus 100 sampled unseen candidates per user. Unrated candidates are not confirmed dislikes.

```python
pd.DataFrame(summary['rating_results']).T.sort_values('rmse')
```

```python
display(Image(filename=ROOT / 'reports' / 'figures' / 'week4_rating_error_comparison.png'))
```

## 2. Precision@10, Recall@10, Hit Rate@10, NDCG@10 and coverage

```python
ranking = pd.DataFrame(summary['ranking_results']).T
ranking.sort_values('precision_at_k', ascending=False)
```

```python
display(Image(filename=ROOT / 'reports' / 'figures' / 'week4_ranking_comparison.png'))
```

## 3. Final selection
Item-Item has the lowest RMSE, while SVD has the strongest Precision@10 and NDCG@10. CineMatch selects SVD because the product objective is ranked movie discovery.

## 5. Interpretation limits
Sampled-negative ranking does not prove that unrated movies are disliked. Offline accuracy also does not measure novelty, diversity, fairness, current availability, or satisfaction in a live product.
